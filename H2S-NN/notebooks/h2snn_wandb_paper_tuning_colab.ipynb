# H2S-NN W&B Paper Tuning - Colab

Fresh-sweep-first primary Bayesian W&B tuning notebook. Node and link tasks use the complete graph with no query-minibatch search. Graph tasks use the fixed implementation batch size of 32 complete graphs and, when SheafCovPool is fixed, search only the post-Gram feature width over {4, 8, 16}.

## 1. Mount Drive And Enter Repo

Edit `PROJECT_DIR` if your repo lives somewhere else.

```python
from pathlib import Path
import os
import sys

IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')
    PROJECT_DIR = Path('/content/drive/MyDrive/H2S-NN_test')
else:
    PROJECT_DIR = Path.cwd()

if not PROJECT_DIR.exists():
    raise FileNotFoundError(f'Project directory not found: {PROJECT_DIR}')

os.chdir(PROJECT_DIR)
print('Working directory:', PROJECT_DIR)
print('Python executable:', sys.executable)
```

## 2. Install Dependencies

This keeps the active Colab PyTorch build and installs PyG wheels for that Torch/CUDA pair.

```python
import importlib.metadata
import os
import signal
import subprocess
import sys

requirements_path = PROJECT_DIR / 'requirements.txt'
if not requirements_path.exists():
    raise FileNotFoundError(f'Missing pinned requirements: {requirements_path}')

try:
    installed_torch = importlib.metadata.version('torch')
except importlib.metadata.PackageNotFoundError:
    installed_torch = ''

target_ready = installed_torch.startswith('2.6.') and '+cu124' in installed_torch
print('installed torch distribution:', installed_torch or 'not installed')
subprocess.check_call([
    sys.executable,
    '-m',
    'pip',
    'install',
    '--upgrade',
    '-r',
    str(requirements_path),
])

if not target_ready:
    print('Pinned Torch/CUDA stack installed. Restarting the runtime; rerun from the first cell.')
    if IN_COLAB:
        os.kill(os.getpid(), signal.SIGKILL)
    raise RuntimeError('Restart Python, then rerun the notebook from the first cell.')
```

## 3. Sanity Check

```python
import subprocess
import sys

import torch
import torch_geometric
import torch_scatter
import torch_sparse

if not torch.__version__.startswith('2.6.'):
    raise RuntimeError(f'Expected torch 2.6.*, found {torch.__version__}. Rerun the install cell.')
if torch.version.cuda != '12.4':
    raise RuntimeError(f'Expected the CUDA 12.4 Torch build, found CUDA {torch.version.cuda}.')

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print('device:', DEVICE)
print('torch:', torch.__version__)
print('torch_geometric:', torch_geometric.__version__)
print('torch_scatter:', torch_scatter.__version__)
print('torch_sparse:', torch_sparse.__version__)
if DEVICE == 'cuda':
    properties = torch.cuda.get_device_properties(0)
    print('GPU:', properties.name)
    print('GPU capacity GiB:', round(properties.total_memory / 1024**3, 2))

subprocess.run([sys.executable, 'scripts/run_h2snn_wandb_tuning.py', '--help'], check=True)
```

## 4. Paper Tuning Knobs

For a fresh sweep, keep `RESUME = False` and `SWEEP_ID = ''`. After a Colab timeout, set `RESUME = True`; the script will reuse the saved sweep id from `STATE_DIR`. Paste `SWEEP_ID` only if you need to attach to a known W&B sweep manually, usually with `NUM_RUNS = 1`.

```python
from pathlib import Path

# Choose one base config. Examples:
#   configs/experiments/acm_h2snn_ampss.yaml       # node classification + AMPSS
#   configs/experiments/dblp_h2snn_ampss.yaml
#   configs/experiments/imdb_h2snn_ampss.yaml
#   configs/experiments/lastfm_h2snn.yaml          # link prediction
#   configs/experiments/mutag_h2snn.yaml           # graph classification
CONFIG_PATH = Path('configs/experiments/acm_h2snn_ampss.yaml')

EXP_NAME = 'acm_h2snn_ampss_node_fullbatch_primary_v5'
OUTPUT_DIR = Path('experiments/colab')
STATE_DIR = OUTPUT_DIR / EXP_NAME / 'tuning_state'

NUM_RUNS = 5
N_TRIALS = 30
EPOCHS = 500
PATIENCE = 50
HYPERBAND_MIN_ITER = 30
HYPERBAND_ETA = 3
D_VALUES = None
SHEAFCOV_FEAT_DIMS = [4, 8, 16]

# Partitioning a campaign across machines: keep NUM_RUNS and BASE_SEED
# fixed everywhere and give each machine a disjoint run range. Never run
# the same run index concurrently.
BASE_SEED = 42
RUN_START = 0
RUN_COUNT = NUM_RUNS

RESUME = False
SWEEP_ID = ''                  # optional manual W&B sweep id; use with NUM_RUNS = 1
WANDB_PROJECT = 'h2snn'
WANDB_ENTITY = ''
WANDB_MODE = 'online'
WANDB_TAGS = ['paper-quality', 'wandb-tuning']

print('Config:', CONFIG_PATH)
print('State dir:', STATE_DIR)
print('After tuning, inspect:', OUTPUT_DIR / EXP_NAME / 'tuning_summary.json')
print('Repeated-run baseline summaries use runs_summary.json from scripts/run_h2snn_runs.py')


import yaml

if not CONFIG_PATH.exists():
    raise FileNotFoundError(f'Base config does not exist: {CONFIG_PATH}')
base_payload = yaml.safe_load(CONFIG_PATH.read_text(encoding='utf-8'))
from src.configs.config import TrainConfig
resolved_base = TrainConfig.from_dict(base_payload)
resolved_task_type = resolved_base.task_type
resolved_pooling_type = resolved_base.data.pooling_type
base_orthogonal_map = str(base_payload.get('model', {}).get('orthogonal_map', 'fasthpp'))
if base_orthogonal_map != 'fasthpp':
    raise ValueError('The current src model and tuning protocol support only orthogonal_map=fasthpp.')
print('Resolved base orthogonal map:', base_orthogonal_map)
print('Resolved task:', resolved_task_type)
print('Requested d values:', D_VALUES or 'safe task defaults')
if resolved_task_type in {'node_classification', 'link_prediction'}:
    print('Training context: complete graph; query minibatching is not enabled')
elif resolved_pooling_type == 'sheafcov':
    print('Raw Gram size: fixed 3 x 3; post-Gram widths:', SHEAFCOV_FEAT_DIMS)
print('Fresh sweep:', not RESUME)
```

## 5. Login To W&B

```python
import os
import wandb

wandb_key = None
if IN_COLAB:
    try:
        from google.colab import userdata
        wandb_key = userdata.get('WANDB_API_KEY')
    except Exception:
        wandb_key = None

if wandb_key:
    wandb.login(key=wandb_key, relogin=True)
else:
    wandb.login()
```

## 6. Run Or Resume Tuning

If Colab stops because of the time limit, reconnect, rerun cells 1-5, then rerun this cell. Completed trials in `STATE_DIR` will be skipped.

```python
import subprocess
import sys

cmd = [
    sys.executable,
    'scripts/run_h2snn_wandb_tuning.py',
    '--config', str(CONFIG_PATH),
    '--exp_name', EXP_NAME,
    '--output_dir', str(OUTPUT_DIR),
    '--device', DEVICE,
    '--num_runs', str(NUM_RUNS),
    '--n_trials', str(N_TRIALS),
    '--epochs', str(EPOCHS),
    '--patience', str(PATIENCE),
    '--hyperband_min_iter', str(HYPERBAND_MIN_ITER),
    '--hyperband_eta', str(HYPERBAND_ETA),
    '--state_dir', str(STATE_DIR),
    '--wandb_project', WANDB_PROJECT,
    '--wandb_mode', WANDB_MODE,
    '--wandb_tags', *WANDB_TAGS,
]
if D_VALUES:
    cmd.extend(['--d_values', *[str(value) for value in D_VALUES]])
if resolved_task_type == 'graph_classification' and resolved_pooling_type == 'sheafcov':
    cmd.extend(['--sheafcov_feat_dims', *[str(value) for value in SHEAFCOV_FEAT_DIMS]])
cmd.extend([
    '--base_seed', str(BASE_SEED),
    '--run_start', str(RUN_START),
    '--run_count', str(RUN_COUNT),
])
if WANDB_ENTITY:
    cmd.extend(['--wandb_entity', WANDB_ENTITY])
if RESUME:
    cmd.append('--resume')
if SWEEP_ID:
    cmd.extend(['--sweep_id', SWEEP_ID])

print('Validating:', ' '.join([*cmd, '--validate_only']))
subprocess.run([*cmd, '--validate_only'], check=True)
print('Starting:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

## 7. Inspect Results

```python
import json

summary_path = OUTPUT_DIR / EXP_NAME / 'tuning_summary.json'
print('summary_path:', summary_path)
if summary_path.exists():
    payload = json.loads(summary_path.read_text(encoding='utf-8'))
    print(json.dumps(payload.get('summary', payload), indent=2)[:4000])
else:
    print('No final tuning summary yet. Check per-run states:')
    for path in sorted(STATE_DIR.glob('tuning_state_run*.json')):
        state = json.loads(path.read_text(encoding='utf-8'))
        print(path.name, 'completed', state.get('completed_trials'), '/', state.get('target_trials'), 'sweep', state.get('sweep_id'))
```
