# H2S-NN Graph Classification W&B Tuning - Colab

Primary W&B tuning notebook for MUTAG/PROTEINS/ENZYMES/NCI1 graph classification. It keeps SheafCovPool fixed, keeps its semantic Gram descriptor fixed at 3 x 3, and searches only the post-Gram feature width over {4, 8, 16}. Each graph remains a complete complex inside the implementation's fixed batch of up to 32 graphs.

```python
from pathlib import Path

BASE_CONFIG_PATH = Path('configs/experiments/mutag_h2snn.yaml')

# AMPSS discovery controls. These are fixed before W&B trials because they define
# the candidate pool and must not vary across trials.
DISCOVERY_MAX_SCHEMA_LENGTH = 2
DISCOVERY_MAX_SCHEMA_CANDIDATES = 128
DISCOVERY_SELECTED_SCHEMA_K = 6
DISCOVERY_MAX_INSTANCES_PER_SCHEMA = 8000
DISCOVERY_MAX_INSTANCES_PER_ANCHOR = 30
DISCOVERY_MAX_TOTAL_INSTANCES = 20000
DISCOVERY_MAX_TOTAL_INCIDENCES = 60000
DISCOVERY_ALLOW_REPEATED_NODE_TYPES = True

EXP_NAME = 'mutag_h2snn_graph_sheafcov_primary_v5'
OUTPUT_DIR = Path('experiments/colab')
STATE_DIR = OUTPUT_DIR / EXP_NAME / 'tuning_state'

NUM_RUNS = 5
N_TRIALS = 50
EPOCHS = 600
PATIENCE = 80
HYPERBAND_MIN_ITER = 30
HYPERBAND_ETA = 3
D_VALUES = [8,16, 32, 64]
SHEAFCOV_FEAT_DIMS = [4, 8, 16]
BASE_SEED = 42
RUN_START = 0
RUN_COUNT = NUM_RUNS

RESUME = False
SWEEP_ID = ''
WANDB_PROJECT = 'h2snn'
WANDB_ENTITY = ''
WANDB_MODE = 'online'
WANDB_TAGS = ['paper-quality', 'graph-classification', 'wandb-tuning']

print('After tuning:', OUTPUT_DIR / EXP_NAME / 'tuning_summary.json')
print('Repeated-run summaries use runs_summary.json from scripts/run_h2snn_runs.py')


import yaml

if not BASE_CONFIG_PATH.exists():
    raise FileNotFoundError(f'Base config does not exist: {BASE_CONFIG_PATH}')
base_payload = yaml.safe_load(BASE_CONFIG_PATH.read_text(encoding='utf-8'))
base_payload.setdefault('discovery', {}).update({
    'max_schema_length': DISCOVERY_MAX_SCHEMA_LENGTH,
    'max_schema_candidates': DISCOVERY_MAX_SCHEMA_CANDIDATES,
    'selected_schema_k': DISCOVERY_SELECTED_SCHEMA_K,
    'max_instances_per_schema': DISCOVERY_MAX_INSTANCES_PER_SCHEMA,
    'max_instances_per_anchor': DISCOVERY_MAX_INSTANCES_PER_ANCHOR,
    'max_total_instances': DISCOVERY_MAX_TOTAL_INSTANCES,
    'max_total_incidences': DISCOVERY_MAX_TOTAL_INCIDENCES,
    'allow_repeated_node_types': DISCOVERY_ALLOW_REPEATED_NODE_TYPES,
})
CONFIG_PATH = OUTPUT_DIR / EXP_NAME / 'notebook_tuning_config.yaml'
CONFIG_PATH.parent.mkdir(parents=True, exist_ok=True)
CONFIG_PATH.write_text(yaml.safe_dump(base_payload, sort_keys=False), encoding='utf-8')
base_orthogonal_map = str(base_payload.get('model', {}).get('orthogonal_map', 'fasthpp'))
if base_orthogonal_map != 'fasthpp':
    raise ValueError('The current src model and tuning protocol support only orthogonal_map=fasthpp.')
base_pooling_type = str(base_payload.get('data', {}).get('pooling_type', ''))
if base_pooling_type not in {'sheafcov', 'sheafconv'}:
    raise ValueError('Post-Gram feature-width tuning requires a SheafCovPool base config.')
print('Resolved base orthogonal map:', base_orthogonal_map)
print('Resolved graph pooling:', base_pooling_type)
print('Requested d values:', D_VALUES or 'safe task defaults')
print('Raw Gram size: fixed 3 x 3 (three semantic views)')
print('Requested post-Gram feature widths:', SHEAFCOV_FEAT_DIMS)
print('Graph batch size: fixed by the implementation at 32; each graph is kept complete')
print('Fresh sweep:', not RESUME)
```

```python
import subprocess
import sys

cmd = [sys.executable, 'scripts/run_h2snn_wandb_tuning.py', '--config', str(CONFIG_PATH), '--exp_name', EXP_NAME, '--output_dir', str(OUTPUT_DIR), '--device', DEVICE, '--num_runs', str(NUM_RUNS), '--n_trials', str(N_TRIALS), '--epochs', str(EPOCHS), '--patience', str(PATIENCE), '--hyperband_min_iter', str(HYPERBAND_MIN_ITER), '--hyperband_eta', str(HYPERBAND_ETA), '--state_dir', str(STATE_DIR), '--wandb_project', WANDB_PROJECT, '--wandb_mode', WANDB_MODE, '--wandb_tags', *WANDB_TAGS]
if D_VALUES:
    cmd.extend(['--d_values', *[str(value) for value in D_VALUES]])
if SHEAFCOV_FEAT_DIMS:
    cmd.extend(['--sheafcov_feat_dims', *[str(value) for value in SHEAFCOV_FEAT_DIMS]])
cmd.extend(['--base_seed', str(BASE_SEED), '--run_start', str(RUN_START), '--run_count', str(RUN_COUNT)])
if WANDB_ENTITY:
    cmd.extend(['--wandb_entity', WANDB_ENTITY])
if RESUME:
    cmd.append('--resume')
if SWEEP_ID:
    cmd.extend(['--sweep_id', SWEEP_ID])
print('Validating:', ' '.join([*cmd, '--validate_only']))
subprocess.run([*cmd, '--validate_only'], check=True)
print('Starting:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

```python
import json
summary_path = OUTPUT_DIR / EXP_NAME / 'tuning_summary.json'
print('summary_path:', summary_path)
if summary_path.exists():
    print(json.dumps(json.loads(summary_path.read_text(encoding='utf-8')).get('summary', {}), indent=2)[:4000])
else:
    for path in sorted(STATE_DIR.glob('tuning_state_run*.json')):
        state = json.loads(path.read_text(encoding='utf-8'))
        print(path.name, 'completed', state.get('completed_trials'), '/', state.get('target_trials'), 'sweep', state.get('sweep_id'))
```
