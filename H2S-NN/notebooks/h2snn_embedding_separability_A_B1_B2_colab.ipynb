# H2S-NN embedding separability: A, B1, and B2

This notebook runs three complementary experiments:

- **A:** fixed class-aligned $O(2)$ oracle diffusion control.
- **B1:** H2S-NN training-time embedding snapshots.
- **B2:** H2S-NN trained once, followed by a longer diffusion-time rollout with learned maps frozen.

A is a positive control, not a learned H2S-NN result. B1 and B2 share one training run when `--experiment all` is selected.

```python
# Colab dependencies.
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'scikit-learn', 'umap-learn', 'pyyaml', 'torch-geometric'], check=True)
```

```python
# Experiment controls. Reduce AVG_NUM_NODES or EPOCHS for a smoke test.
EXPERIMENT = 'all'
METHOD = 'all'
DEVICE = 'cuda'
EPOCHS = 500
SNAPSHOT_EVERY = 5
AVG_NUM_NODES = 400
ROLLOUT_STEPS = 8
ORACLE_STEPS = 14
OUTPUT_DIR = Path('figures/embedding_separability')
command = [
    sys.executable, '-u', 'scripts/run_embedding_separability_experiments.py',
    '--experiment', EXPERIMENT, '--method', METHOD, '--device', DEVICE,
    '--epochs', str(EPOCHS), '--snapshot-every', str(SNAPSHOT_EVERY),
    '--avg-num-nodes', str(AVG_NUM_NODES), '--rollout-steps', str(ROLLOUT_STEPS),
    '--oracle-steps', str(ORACLE_STEPS), '--output-dir', str(OUTPUT_DIR),
    '--seed', '42', '--hidden-dim', '16', '--diffusion-steps', '3','--point-size', '12',
]
print('Running:', ' '.join(command))
subprocess.run(command, check=True)
```
